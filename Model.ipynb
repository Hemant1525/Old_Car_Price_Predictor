# TRAIN_TEST_SPLIT

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.svm import SVR
from sklearn.neighbors import KNeighborsRegressor
from sklearn.ensemble import RandomForestRegressor
```

```python
from sklearn.metrics import mean_squared_error,mean_absolute_error
```

```python
lr = LinearRegression()
lr.fit(x_train,y_train)
lr.score(x_train,y_train)*100,lr.score(x_test,y_test)*100
```

```python
mean_squared_error(y_test,lr.predict(x_test)),mean_absolute_error(y_test,lr.predict(x_test))
```

```python
dt = DecisionTreeRegressor()
dt.fit(x_train,y_train)
dt.score(x_train,y_train)*100,dt.score(x_test,y_test)*100
```

```python
mean_squared_error(y_test,dt.predict(x_test)),mean_absolute_error(y_test,dt.predict(x_test))
```

```python
rf = RandomForestRegressor(n_estimators=100)
rf.fit(x_train,y_train)
rf.score(x_train,y_train)*100,rf.score(x_test,y_test)*100
```

```python
mean_squared_error(y_test,rf.predict(x_test)),mean_absolute_error(y_test,rf.predict(x_test))
```

```python
# Actual vs Predicted scatter — sabse important graph model ke liye
y_pred = rf.predict(x_test)

plt.figure(figsize=(7,7))
plt.scatter(y_test, y_pred, alpha=0.4)
plt.plot([y_test.min(), y_test.max()], [y_test.min(), y_test.max()], 'r--', lw=2)
plt.xlabel("Actual Price (₹)")
plt.ylabel("Predicted Price (₹)")
plt.title("Actual vs Predicted Price")
plt.show()
```

```python
# Residual plot — errors kahan zyada hain, ye dikhata hai
residuals = y_test - y_pred

plt.figure(figsize=(8,5))
sns.scatterplot(x=y_pred, y=residuals, alpha=0.4)
plt.axhline(0, color='red', linestyle='--')
plt.xlabel("Predicted Price (₹)")
plt.ylabel("Residual (Actual - Predicted)")
plt.title("Residual Plot")
plt.show()
```

```python
# Feature importance — kaunsa feature sabse zyada matter karta hai
importance_df = pd.DataFrame({
    'feature': x.columns,
    'importance': rf.feature_importances_
}).sort_values('importance', ascending=False)

plt.figure(figsize=(8,5))
sns.barplot(data=importance_df, x='importance', y='feature')
plt.title("Feature Importance (Random Forest)")
plt.show()
```

```python
# Model comparison — lr, dt, rf teeno ka R2 side by side
model_scores = pd.DataFrame({
    'Model': ['Linear Regression', 'Decision Tree', 'Random Forest'],
    'Train R2': [lr.score(x_train,y_train)*100, dt.score(x_train,y_train)*100, rf.score(x_train,y_train)*100],
    'Test R2':  [lr.score(x_test,y_test)*100, dt.score(x_test,y_test)*100, rf.score(x_test,y_test)*100]
})

model_scores.set_index('Model').plot(kind='bar', figsize=(8,5))
plt.title("Model Comparison (Train vs Test R2)")
plt.ylabel("R2 Score (%)")
plt.xticks(rotation=0)
plt.show()
```
